### Skeleton GT Review

Max-Z MIPs of the debug TIFFs dumped by `biapy_work_folder/check_skeleton_channel.py`
(`skeleton_channel_check` and `skeleton_channel_check_touch`).

```python
import matplotlib.pyplot as plt
import numpy as np
import tifffile

from view_augments.base_data import enhance_display
from view_augments.metric_review import mip_biapy_gt_instance

SKEL_CHECK_DIRS = [
    Path("biapy_work_folder/skeleton_channel_check"),
    Path("biapy_work_folder/skeleton_channel_check_touch"),
]

STEMS = {
    "labels": "vol_original_labels",
    "F": "vol_F_foreground",
    "Db": "vol_Db_distance_to_border",
    "Sk": "vol_Sk_tubed_skeleton",
}


def max_z_mip(path):
    """Load a 3D TIFF and return its max-Z MIP (Y, X)."""
    vol = np.asarray(tifffile.imread(path))
    if vol.ndim == 4 and vol.shape[-1] == 1:
        vol = vol[..., 0]
    if vol.ndim != 3:
        raise ValueError(f"Expected 3D volume in {path}, got {vol.shape}")
    return vol.max(axis=0)


def _resolve_stem(d, stem):
    for ext in (".tif", ".tiff"):
        p = d / f"{stem}{ext}"
        if p.is_file():
            return p
    raise FileNotFoundError(f"Missing {stem}.tif(f) in {d}")


def plot_skeleton_check_mips(dirs, figsize_cell=(4, 4), dpi=120):
    """2x3 max-Z MIPs: labels | F | Db / Sk | F+Sk overlay | Sk\\F residual."""
    for d in dirs:
        d = Path(d)
        if not d.is_dir():
            print(f"Missing dir {d}")
            continue

        paths = {k: _resolve_stem(d, stem) for k, stem in STEMS.items()}
        labels_vol = np.asarray(tifffile.imread(paths["labels"]))
        if labels_vol.ndim == 4 and labels_vol.shape[-1] == 1:
            labels_vol = labels_vol[..., 0]

        f_mip = max_z_mip(paths["F"]) > 0
        sk_mip = max_z_mip(paths["Sk"]) > 0
        db_mip = max_z_mip(paths["Db"])
        residual = sk_mip & ~f_mip

        sk_n = int(sk_mip.sum())
        f_n = int(f_mip.sum())
        sk_of_f = 100.0 * sk_n / f_n if f_n else 0.0

        # Overlay: F gray, Sk red on black.
        overlay = np.zeros((*f_mip.shape, 3), dtype=np.float32)
        overlay[..., 0] = np.maximum(f_mip.astype(np.float32) * 0.35, sk_mip.astype(np.float32))
        overlay[..., 1] = f_mip.astype(np.float32) * 0.35
        overlay[..., 2] = f_mip.astype(np.float32) * 0.35

        residual_rgb = np.zeros((*f_mip.shape, 3), dtype=np.float32)
        residual_rgb[..., 0] = residual.astype(np.float32)

        fig, axes = plt.subplots(
            2, 3, figsize=(figsize_cell[0] * 3, figsize_cell[1] * 2), dpi=dpi
        )
        fig.suptitle(str(d), fontsize=11)

        mip_biapy_gt_instance(
            axes[0, 0], labels_vol, paths["labels"].stem, z_axis=0, title="original_labels"
        )

        axes[0, 1].imshow(f_mip.astype(np.float32), cmap="hot", interpolation="nearest")
        axes[0, 1].set_title("F foreground", fontsize=8)
        axes[0, 1].axis("off")

        axes[0, 2].imshow(enhance_display(db_mip), cmap="viridis", interpolation="nearest")
        axes[0, 2].set_title("Db distance", fontsize=8)
        axes[0, 2].axis("off")

        axes[1, 0].imshow(sk_mip.astype(np.float32), cmap="hot", interpolation="nearest")
        axes[1, 0].set_title("Sk tubed skeleton", fontsize=8)
        axes[1, 0].axis("off")

        axes[1, 1].imshow(overlay, interpolation="nearest")
        axes[1, 1].set_title(f"F+Sk overlay (Sk={sk_n}, {sk_of_f:.1f}% of F)", fontsize=8)
        axes[1, 1].axis("off")

        axes[1, 2].imshow(residual_rgb, interpolation="nearest")
        axes[1, 2].set_title(f"Sk\\\\F residual (n={int(residual.sum())})", fontsize=8)
        axes[1, 2].axis("off")

        plt.tight_layout()
        plt.show()


plot_skeleton_check_mips(SKEL_CHECK_DIRS)
```
