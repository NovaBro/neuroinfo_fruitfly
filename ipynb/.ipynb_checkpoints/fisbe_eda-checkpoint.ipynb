## Create tiff Version for BiaPy

````python
import zarr
import shutil
from tifffile import imwrite
from tqdm import tqdm


# Create x_y_dir for BiaPy
def create_x_y_dir(path:Path):
    (path / 'raw').mkdir(parents=True, exist_ok=True)
    (path / 'label').mkdir(parents=True, exist_ok=True)

def create_dirs(root_dir:Path, dirs: list[Path]):
    for dir in dirs:
        (root_dir / dir).mkdir(parents=True, exist_ok=True)
        create_x_y_dir(root_dir / dir)

TIFF_PATH = Path('./fisbe/biapy').resolve()
shutil.rmtree(TIFF_PATH, ignore_errors=True)
TIFF_PATH.mkdir(parents=True, exist_ok=True)
create_dirs(TIFF_PATH, ['train', 'test', 'val'])
print(DATASET)
print(TIFF_PATH)
````

````python
raw = zarr.open(DATASET /  'train' / 'R38F04-20181005_63_G3.zarr', mode='r', path="volumes/raw")
seg = zarr.open(DATASET /  'train' / 'R38F04-20181005_63_G3.zarr', mode='r', path="volumes/gt_instances")
raw_np = np.array(raw)
seg_np = np.array(seg)
raw_np.shape, seg_np.shape, np.unique(seg_np)
````

````python
train_files = os.listdir(train_path)
test_files = os.listdir(test_path)
val_files = os.listdir(val_path)

def create_tiff_version(data_path, tiff_path=TIFF_PATH):
    for file in tqdm(os.listdir(data_path)):
        raw = zarr.open(data_path / file, mode='r', path="volumes/raw")
        seg = zarr.open(data_path / file, mode='r', path="volumes/gt_instances")
        raw_np = np.array(raw)
        seg_np = np.array(seg)  
        raw_np.shape, seg_np.shape, np.unique(seg_np)
        imwrite(tiff_path / 'raw' / f"{file}.tiff", raw_np)
        imwrite(tiff_path / 'label' / f"{file}_seg.tiff", seg_np)

# Create tiff version for BiaPy
for split in ['train', 'test', 'val']:
    tqdm.write(f"Running Split {split}")
    create_tiff_version(DATASET / split, TIFF_PATH / split)
````

## View Results of PatchPerPix on FISBe

````python
import os
from pathlib import Path

# If the kernel starts in .../AlmaAI/ipynb
root = Path.cwd()
if root.name == "ipynb":
    root = root.parent
os.chdir(root)

import zarr
import shutil
from tifffile import imwrite
from tqdm import tqdm
import numpy as np
import matplotlib.pyplot as plt
````

````python
zarr_data1_code = zarr.open("PatchPerPix/experiments/ppp_experiments/flylight_setup01_260610_113923_103163/val/processed/100/VT033614-20171124_64_H1.zarr", mode='r', path="volumes/pred_code")
print("Orignal Code Shape:", zarr_data1_code.shape)
subset_code = zarr_data1_code[100]
subset_code = np.array(subset_code)
print("Subset Code Shape:", subset_code.shape)

zarr_data1_numinst = zarr.open("PatchPerPix/experiments/ppp_experiments/flylight_setup01_260610_113923_103163/val/processed/100/VT033614-20171124_64_H1.zarr", mode='r', path="volumes/pred_numinst")
print("Orignal Numinst Shape:", zarr_data1_numinst.shape)
subset_numinst = zarr_data1_numinst[0]
subset_numinst = np.array(subset_numinst)
print("Subset Numinst Shape:", subset_numinst.shape)



# zarr_data2_code = zarr.open("PatchPerPix/experiments/ppp_experiments/flylight_setup01_260610_113923_103163/val/processed/100/VT033614-20171124_64_H5.zarr", mode='r', path="volumes/pred_code")
# zarr_data2_numinst = zarr.open("PatchPerPix/experiments/ppp_experiments/flylight_setup01_260610_113923_103163/val/processed/100/VT033614-20171124_64_H5.zarr", mode='r', path="volumes/pred_numinst")
# np.array(zarr_data1_numinst)

````

````python
SAMPLE = "VT033614-20171124_64_H1"
CODE_IDX = 100
NUMINST_THRESHS = [0.9, 0.1]  # from flylight setup01 config

FISBE_SAMPLE = Path(f"fisbe/completely/val/{SAMPLE}.zarr")
raw_zarr = zarr.open(FISBE_SAMPLE, mode="r", path="volumes/raw_normalized")
gt_instances_zarr = zarr.open(FISBE_SAMPLE, mode="r", path="volumes/gt_instances_rm_5")
gt_numinst_zarr = zarr.open(FISBE_SAMPLE, mode="r", path="volumes/gt_numinst")

pred_numinst_vol = np.array(zarr_data1_numinst)


def raw_rgb_slice(raw, z):
    slc = np.array(raw[:, z], dtype=np.float32)
    slc = (slc - slc.min()) / (np.ptp(slc) + 1e-8)
    return np.moveaxis(slc, 0, -1)


def pred_numinst_count(numinst, threshs=None):
    numinst = np.asarray(numinst)
    if threshs is not None:
        out = np.zeros(numinst.shape[1:], dtype=np.uint8)
        for i, thresh in enumerate(threshs):
            out[numinst[i + 1] > thresh] = i + 1
        return out
    return np.argmax(numinst, axis=0).astype(np.uint8)


def gt_label_slice(gt_instances, z):
    merged = np.zeros(gt_instances.shape[1:], dtype=np.uint16)
    offset = 0
    for ch in range(gt_instances.shape[0]):
        labels = np.array(gt_instances[ch, z])
        if labels.max() == 0:
            continue
        mask = labels > 0
        merged[mask] = labels[mask] + offset
        offset += int(labels.max())
    return merged


def show_ppp_slice(raw, code, numinst, gt_instances, gt_numinst, z, title_prefix=""):
    pred_count = pred_numinst_count(numinst, NUMINST_THRESHS)
    code_vmin, code_vmax = np.percentile(code, [1, 99])

    fig, axes = plt.subplots(2, 4, figsize=(16, 8))

    axes[0, 0].imshow(raw_rgb_slice(raw, z))
    axes[0, 0].set_title(f"{title_prefix}raw RGB (z={z})")
    axes[0, 0].axis("off")

    axes[0, 1].imshow(code[z], cmap="viridis", vmin=code_vmin, vmax=code_vmax)
    axes[0, 1].set_title(f"code unit {CODE_IDX}")
    axes[0, 1].axis("off")

    axes[0, 2].imshow(numinst[0, z], cmap="magma", vmin=0, vmax=1)
    axes[0, 2].set_title("numinst ch0 (0 instances)")
    axes[0, 2].axis("off")

    axes[0, 3].imshow(numinst[1, z], cmap="magma", vmin=0, vmax=1)
    axes[0, 3].set_title("numinst ch1 (1 instance)")
    axes[0, 3].axis("off")

    axes[1, 0].imshow(numinst[2, z], cmap="magma", vmin=0, vmax=1)
    axes[1, 0].set_title("numinst ch2 (2+ instances)")
    axes[1, 0].axis("off")

    axes[1, 1].imshow(pred_count[z], cmap="tab10", vmin=0, vmax=2)
    axes[1, 1].set_title("predicted instance count")
    axes[1, 1].axis("off")

    axes[1, 2].imshow(np.array(gt_numinst[z]), cmap="tab10", vmin=0, vmax=2)
    axes[1, 2].set_title("GT numinst")
    axes[1, 2].axis("off")

    gt_slice = gt_label_slice(gt_instances, z)
    axes[1, 3].imshow(gt_slice, cmap="nipy_spectral")
    axes[1, 3].set_title(f"GT instances ({len(np.unique(gt_slice)) - 1} objects)")
    axes[1, 3].axis("off")

    plt.tight_layout()
    plt.show()


def mip_rgb(raw, axis=0):
    mip = np.array(raw, dtype=np.float32).max(axis=axis + 1)
    mip = (mip - mip.min()) / (np.ptp(mip) + 1e-8)
    return np.moveaxis(mip, 0, -1)


def show_ppp_mip(raw, code, numinst, gt_instances, gt_numinst, title_prefix="", axis=0):
    pred_count = pred_numinst_count(numinst, NUMINST_THRESHS)
    code_mip = code.max(axis=axis)
    numinst_mip = numinst.max(axis=axis + 1)
    pred_count_mip = pred_count.max(axis=axis)
    gt_numinst_mip = np.array(gt_numinst).max(axis=axis)

    gt_mip = np.zeros(gt_instances.shape[1:], dtype=np.uint16)
    offset = 0
    for ch in range(gt_instances.shape[0]):
        labels = np.array(gt_instances[ch]).max(axis=axis)
        if labels.max() == 0:
            continue
        mask = labels > 0
        gt_mip[mask] = labels[mask] + offset
        offset += int(labels.max())

    code_vmin, code_vmax = np.percentile(code_mip, [1, 99])
    fig, axes = plt.subplots(2, 4, figsize=(16, 8))

    axes[0, 0].imshow(mip_rgb(raw, axis=axis))
    axes[0, 0].set_title(f"{title_prefix}raw RGB MIP")
    axes[0, 0].axis("off")

    axes[0, 1].imshow(code_mip, cmap="viridis", vmin=code_vmin, vmax=code_vmax)
    axes[0, 1].set_title(f"code unit {CODE_IDX} MIP")
    axes[0, 1].axis("off")

    axes[0, 2].imshow(numinst_mip[0], cmap="magma", vmin=0, vmax=1)
    axes[0, 2].set_title("numinst ch0 MIP")
    axes[0, 2].axis("off")

    axes[0, 3].imshow(numinst_mip[1], cmap="magma", vmin=0, vmax=1)
    axes[0, 3].set_title("numinst ch1 MIP")
    axes[0, 3].axis("off")

    axes[1, 0].imshow(numinst_mip[2], cmap="magma", vmin=0, vmax=1)
    axes[1, 0].set_title("numinst ch2 MIP")
    axes[1, 0].axis("off")

    axes[1, 1].imshow(pred_count_mip, cmap="tab10", vmin=0, vmax=2)
    axes[1, 1].set_title("predicted instance count MIP")
    axes[1, 1].axis("off")

    axes[1, 2].imshow(gt_numinst_mip, cmap="tab10", vmin=0, vmax=2)
    axes[1, 2].set_title("GT numinst MIP")
    axes[1, 2].axis("off")

    axes[1, 3].imshow(gt_mip, cmap="nipy_spectral")
    axes[1, 3].set_title(f"GT instances MIP ({len(np.unique(gt_mip)) - 1} labels)")
    axes[1, 3].axis("off")

    plt.tight_layout()
    plt.show()


z_mid = subset_code.shape[0] // 2
show_ppp_slice(
    raw_zarr,
    subset_code,
    pred_numinst_vol,
    gt_instances_zarr,
    gt_numinst_zarr,
    z_mid,
    title_prefix=f"{SAMPLE} | ",
)
show_ppp_mip(
    raw_zarr,
    subset_code,
    pred_numinst_vol,
    gt_instances_zarr,
    gt_numinst_zarr,
    title_prefix=f"{SAMPLE} | ",
)
````

## View Results of BiaPy on FISBe

BiaPy test outputs for the 7 volumes in `fisbe/biapy/test/raw/`:

- `per_image/` — model raw output: 2 probability channels (**F** = foreground, **C** = contour), shape `(Z, 2, Y, X)`, float32 in `[0, 1]`
- `per_image_instances/` — final instance segmentation from watershed on those channels, shape `(Z, Y, X)`, uint16 label IDs



````python
import os
from pathlib import Path

# If the kernel starts in .../ipynb
root = Path.cwd()
if root.name == "ipynb":
    root = root.parent
os.chdir(root)

import zarr
import shutil
import tifffile
from tifffile import imwrite
from tqdm import tqdm
import numpy as np
import matplotlib.pyplot as plt
````

````python
BIAPY_RESULT_ROOT = Path("Biapy/results/3d_instance_segmentation/results/3d_instance_segmentation_1")
BIAPY_PER_IMAGE = BIAPY_RESULT_ROOT / "per_image"
BIAPY_PER_IMAGE_INSTANCES = BIAPY_RESULT_ROOT / "per_image_instances"
````

````python
TEST_RAW_DIR = Path("fisbe/biapy/test/raw")

def sample_stem(path: Path) -> str:
    name = path.name
    for suffix in (".zarr.tif", ".zarr.tiff", ".tif", ".tiff"):
        if name.endswith(suffix):
            return name[: -len(suffix)]
    return path.stem

def load_biapy_test_sample(stem: str):
    """Load raw test volume plus matching BiaPy per_image and instance outputs."""
    raw_path = next(TEST_RAW_DIR.glob(f"{stem}.zarr.tiff"))
    prob_path = BIAPY_PER_IMAGE / f"{stem}.zarr.tif"
    inst_path = BIAPY_PER_IMAGE_INSTANCES / f"{stem}.zarr.tif"

    raw = tifffile.imread(raw_path)          # (Z, C, Y, X)
    probs = tifffile.imread(prob_path)       # (Z, 2, Y, X)
    instances = tifffile.imread(inst_path)   # (Z, Y, X)

    return raw, probs, instances

sample_names = sorted(sample_stem(p) for p in BIAPY_PER_IMAGE.glob("*.tif"))
print(f"{len(sample_names)} test volumes with BiaPy outputs:")
for name in sample_names:
    print(" ", name)

SAMPLE = sample_names[0]
raw_vol, prob_vol, inst_vol = load_biapy_test_sample(SAMPLE)
print(f"\nSelected: {SAMPLE}")
print("raw:", raw_vol.shape, raw_vol.dtype)
print("probs:", prob_vol.shape, prob_vol.dtype)
print("instances:", inst_vol.shape, inst_vol.dtype, "labels:", len(np.unique(inst_vol)))
````

````python
from IPython.display import HTML, display
from matplotlib.animation import FuncAnimation

def raw_rgb_slice(raw, z):
    """Composite RGB channels for one z-slice."""
    slc = raw[z].astype(np.float32)
    slc = (slc - slc.min()) / (np.ptp(slc) + 1e-8)
    return np.moveaxis(slc, 0, -1)

def label_slice(instances, z):
    return instances[z]

def show_biapy_slice(raw, probs, instances, z, title_prefix=""):
    fig, axes = plt.subplots(1, 4, figsize=(16, 4))

    axes[0].imshow(raw_rgb_slice(raw, z))
    axes[0].set_title(f"{title_prefix}raw RGB (z={z})")
    axes[0].axis("off")

    axes[1].imshow(probs[z, 0], cmap="magma", vmin=0, vmax=1)
    axes[1].set_title("F channel (foreground)")
    axes[1].axis("off")

    axes[2].imshow(probs[z, 1], cmap="magma", vmin=0, vmax=1)
    axes[2].set_title("C channel (contour)")
    axes[2].axis("off")

    inst = label_slice(instances, z)
    axes[3].imshow(inst, cmap="nipy_spectral")
    axes[3].set_title(f"instances ({len(np.unique(inst)) - 1} objects)")
    axes[3].axis("off")

    plt.tight_layout()
    plt.show()

z_mid = raw_vol.shape[0] // 2
show_biapy_slice(raw_vol, prob_vol, inst_vol, z_mid, title_prefix=f"{SAMPLE} | ")

def animate_biapy_volume(raw, probs, instances, title_prefix="", interval=100):
    z_slices = raw.shape[0]

    fig, axes = plt.subplots(1, 4, figsize=(16, 4))
    rgb_im = axes[0].imshow(raw_rgb_slice(raw, 0))
    f_im = axes[1].imshow(probs[0, 0], cmap="magma", vmin=0, vmax=1)
    c_im = axes[2].imshow(probs[0, 1], cmap="magma", vmin=0, vmax=1)
    inst_im = axes[3].imshow(label_slice(instances, 0), cmap="nipy_spectral")

    for ax, name in zip(axes, ["raw RGB", "F channel", "C channel", "instances"]):
        ax.set_title(name)
        ax.axis("off")

    def update(frame):
        rgb_im.set_array(raw_rgb_slice(raw, frame))
        f_im.set_array(probs[frame, 0])
        c_im.set_array(probs[frame, 1])
        inst_im.set_array(label_slice(instances, frame))
        fig.suptitle(f"{title_prefix}z = {frame}/{z_slices - 1}", y=1.02)
        return [rgb_im, f_im, c_im, inst_im]

    ani = FuncAnimation(fig, update, frames=z_slices, interval=interval, blit=False)
    plt.close(fig)
    display(HTML(ani.to_jshtml()))

animate_biapy_volume(raw_vol, prob_vol, inst_vol, title_prefix=f"{SAMPLE} | ")
````

````python
def mip_rgb(raw, axis=0):
    """Max-intensity projection of RGB channels along the stack axis."""
    mip = raw.astype(np.float32).max(axis=axis)  # (C, Y, X)
    mip = (mip - mip.min()) / (np.ptp(mip) + 1e-8)
    return np.moveaxis(mip, 0, -1)

def show_biapy_mip(raw, probs, instances, title_prefix="", axis=0):
    """4-panel max-intensity projection along axis (default Z)."""
    prob_mip = probs.max(axis=axis)
    inst_mip = instances.max(axis=axis)

    fig, axes = plt.subplots(1, 4, figsize=(16, 4))

    axes[0].imshow(mip_rgb(raw, axis=axis))
    axes[0].set_title(f"{title_prefix}raw RGB MIP")
    axes[0].axis("off")

    axes[1].imshow(prob_mip[0], cmap="magma", vmin=0, vmax=1)
    axes[1].set_title("F channel MIP")
    axes[1].axis("off")

    axes[2].imshow(prob_mip[1], cmap="magma", vmin=0, vmax=1)
    axes[2].set_title("C channel MIP")
    axes[2].axis("off")

    axes[3].imshow(inst_mip, cmap="nipy_spectral")
    axes[3].set_title(f"instances MIP ({len(np.unique(inst_mip)) - 1} labels)")
    axes[3].axis("off")

    plt.suptitle(f"Maximum intensity projection (axis={axis}, depth={raw.shape[axis]} slices)", y=1.02)
    plt.tight_layout()
    plt.savefig(f"fisbe/biapy/results/3d_instance_segmentation/results/3d_instance_segmentation_1/mip_{title_prefix}.png")
    plt.show()

show_biapy_mip(raw_vol, prob_vol, inst_vol, title_prefix=f"{SAMPLE} | ")
````
