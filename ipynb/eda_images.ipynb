```python
## Image viewing
import os
from pathlib import Path

# Force PyImageJ to use Conda's Java, ignoring system defaults.
# Must be set BEFORE scyjava/imagej is imported (those trigger JVM startup).
if "CONDA_PREFIX" in os.environ:
    os.environ["JAVA_HOME"] = os.environ["CONDA_PREFIX"]

# Ensure the JVM is started in headless mode from the very first call.
# In a Jupyter kernel the JVM only starts once; if it comes up without the
# right options, the IJ1 legacy layer ends up 'Inactive' and IJ.openImage()
# (which the H5J_Loader_Plugin relies on) silently returns null.
import scyjava
scyjava.config.add_option("-Djava.awt.headless=true")

import imagej
import numpy as np

# FIJI_APP = "/home/william-zheng/Downloads/Fiji.app"
# PROJECT_DIR = "/home/william-zheng/Documents/Programming/Python/NeuroInformatics/summer_2026/neuroinfo_fruitfly"
FIJI_APP = "/Users/vuhepola/Desktop/Fiji"
PROJECT_DIR = Path.cwd()
DATA_DIR = PROJECT_DIR / "data"

print(FIJI_APP)
print(PROJECT_DIR)
print(DATA_DIR)
```

```python
print("Initializing headless Fiji environment...")
ij = imagej.init(FIJI_APP, mode="headless", add_legacy=True)
# ij = imagej.init()
# ij.getVersion() returns a java.lang.String; coerce to a Python str.
version = str(ij.getVersion())
print(f"ImageJ version: {version}")
if version.endswith("/Inactive"):
    raise RuntimeError(
        "Legacy ImageJ1 is Inactive in this kernel. Restart the Jupyter "
        "kernel (Kernel -> Restart) and rerun this cell from a clean state."
    )
    
IJ = scyjava.jimport("ij.IJ")
```

```python
file_path = os.path.join(
    DATA_DIR,
    # "R78H12-20190409_62_D3-m-40x-central-GAL4-unaligned_stack.h5j",
    "SS04748-20170324_20_I1-f-20x-ventral_nerve_cord-Split_GAL4-unaligned_stack.h5j",
    # "downloads/Split-GAL4 Omnibus Rescreen/SS02951/SS02951-20230317_43_D5-f-20x-brain-Split_GAL4-JRC2018_Unisex_20x_HR-aligned_stack.h5j",
    # "downloads/Split-GAL4 Omnibus Rescreen/SS02951/SS02951-20230317_43_D5-f-20x-brain-Split_GAL4-JRC2018_Unisex_20x_HR-aligned_stack.h5j"
    # "downloads/Descending Neurons 2018/SS01076/SS01076-20140623_33_J1-f-20x-ventral_nerve_cord-Split_GAL4-unaligned_stack.h5j"
)
if not os.path.isfile(file_path):
    raise FileNotFoundError(file_path)

# The H5J format is Janelia's HDF5 + JPEG2000 container. It is handled by
# the H5J_Loader_Plugin (an IJ1 plugin), not by the SciJava IO service.
# Route through the legacy IJ.openImage so the plugin is invoked.
print(f"Opening {file_path}...")
imp = IJ.openImage(file_path)
if imp is None:
    raise RuntimeError(
        f"IJ.openImage returned null for {file_path}. "
        "Check that the H5J_Loader_Plugin jar is present in Fiji.app/plugins."
    )

print(f"Opened ImagePlus: title={imp.getTitle()}, "
      f"dims={imp.getWidth()}x{imp.getHeight()}x{imp.getNSlices()} "
      f"channels={imp.getNChannels()} frames={imp.getNFrames()}")

print("Converting ImagePlus to NumPy array...")
np_image = ij.py.from_java(imp)
if hasattr(np_image, "values"):
    np_image = np_image.values

print("Extraction successful!")
print(f"Array Type: {type(np_image)}")
print(f"Data Type (dtype): {np_image.dtype}")
print(f"Array Shape: {np_image.shape}")
```

```python
import matplotlib.pyplot as plt
import matplotlib.animation as animation
from IPython.display import HTML
import numpy as np

# 0.0 keeps the inversion as-is; higher values progressively dim near-white
# pixels (those where all RGB channels are high) while leaving saturated
# colours mostly untouched. 1.0 turns pure white fully black.
white_dampen = 0.5

stack = np.asarray(np_image)
print(stack.shape)

print(stack.shape)

if stack.ndim == 4:
    n_channels = stack.shape[-1]
    if n_channels < 3:
        pad_width = [(0, 0)] * 3 + [(3 - n_channels, 0)]
        stack = np.pad(stack, pad_width, mode="constant")
elif stack.ndim != 3:
    raise ValueError(f"Expected a (Z,Y,X) or (Z,Y,X,C) array, got {stack.shape}")

# Normalize to uint8 so matplotlib can display RGB/RGBA directly
# (and a grayscale stack stretched into a sensible range for cmap).
p_lo, p_hi = np.percentile(stack, (1, 99.5))
stack = np.clip((stack - p_lo) / max(p_hi - p_lo, 1e-9), 0, 1)
stack = (stack * 255).astype(np.uint8)

# # Invert the colour channels so structures that previously read as black
# # render as white. For RGBA, keep the original alpha channel untouched so
# # transparency is preserved against the black axes background.
# if stack.ndim == 4 and stack.shape[-1] == 4:
#     rgb = 255 - stack[..., :3]
#     alpha = stack[..., 3:4]
#     stack = np.concatenate([rgb, alpha], axis=-1)
# elif stack.ndim == 4 and stack.shape[-1] == 3:
#     stack = 255 - stack

# # Suppress unsaturated/whitish pixels. The "white component" of a pixel is
# # min(R, G, B); subtracting a fraction of it darkens whites uniformly while
# # leaving coloured pixels (where one channel is much lower) mostly intact.
# if stack.ndim == 4 and 0.0 < white_dampen <= 1.0:
#     rgb = stack[..., :3].astype(np.int16)
#     white = rgb.min(axis=-1, keepdims=True)
#     rgb = np.clip(rgb - (white * white_dampen).astype(np.int16), 0, 255).astype(np.uint8)
#     if stack.shape[-1] == 4:
#         stack = np.concatenate([rgb, stack[..., 3:4]], axis=-1)
#     else:
#         stack = rgb

n_slices = stack.shape[0]
is_rgb = stack.ndim == 4
# For grayscale, use "gray" (high=white) so the convention matches the RGBA
# inversion above; the black axes background then forms the dark backdrop.
imshow_kwargs = {} if is_rgb else {"cmap": "gray", "vmin": 0, "vmax": 255}

fig, ax = plt.subplots()
fig.patch.set_facecolor("black")
ax.set_facecolor("black")
ax.axis("off")
im = ax.imshow(stack[0], **imshow_kwargs)
title = ax.set_title(f"Slice 1/{n_slices}", color="white")

def update(frame):
    im.set_data(stack[frame])
    title.set_text(f"Slice {frame + 1}/{n_slices}")
    return [im, title]

ani = animation.FuncAnimation(
    fig, update, frames=n_slices, interval=100, blit=False, repeat=True
)

plt.close(fig)
HTML(ani.to_jshtml())
```

## Skeletonization Animation Viz


```python
import tifffile as tiff
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import animation
from IPython.display import HTML
from pathlib import Path

# Load skeletal stack from TIFF
# skeleton_tif_path = "R78H12-20190409_62_D3-m-40x-central-GAL4-unaligned_stack_skeleton.tif"
name = Path(file_path).name.replace(".h5j", "_skeleton.tif")
skeleton_tif_path = Path('skeletons') / name
skeleton_stack = tiff.imread(skeleton_tif_path)  # shape: (Z, Y, X) or (Z, Y, X, C)

# Normalize for display as uint8, stretch to percentiles
p_lo, p_hi = np.percentile(skeleton_stack, (1, 99.5))
norm_stack = np.clip((skeleton_stack - p_lo) / max(p_hi - p_lo, 1e-9), 0, 1)
norm_stack = (norm_stack * 255).astype(np.uint8)

n_slices = norm_stack.shape[0]
is_rgb = norm_stack.ndim == 4
imshow_kwargs = {} if is_rgb else {"cmap": "gray_r", "vmin": 0, "vmax": 255}

fig, ax = plt.subplots()
ax.axis("off")
im = ax.imshow(norm_stack[0], **imshow_kwargs)
title = ax.set_title(f"Skeleton Slice 1/{n_slices}")

def update_skel(frame):
    im.set_data(norm_stack[frame])
    title.set_text(f"Skeleton Slice {frame + 1}/{n_slices}")
    return [im, title]

ani_skel = animation.FuncAnimation(
    fig, update_skel, frames=n_slices, interval=100, blit=False, repeat=True
)

plt.close(fig)
HTML(ani_skel.to_jshtml())
```
